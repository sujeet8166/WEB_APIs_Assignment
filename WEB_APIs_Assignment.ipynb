Q1. What is an API? Give an example, where an API is used in real life.

Ans:- API (Application Programming Interface) ek set of rules aur protocols hai jo do software applications ko ek‑dusre se interact karne deta hai,
Example: Jab tum Zomato app use karte ho aur restaurant ka location Google Maps pe dekhte ho, Zomato internally Google Maps API use karta hai.

Q2. Give advantages and disadvantages of using API.

Ans:- Advantages: Easy integration between different systems, Saves development time, Provides security by controlling access, Encourages reusability.
Disadvantages: Dependency on third‑party services, Version changes can break compatibility, Security risks if not implemented properly, Limited customization.

Q3. What is a Web API? Differentiate between API and Web API.

Ans:- Web API ek API hai jo specifically web protocols (HTTP/HTTPS) ke through access hoti hai. Ye client aur server ke beech communication ke liye use hoti hai.
Difference: API: General interface jo kisi bhi application ke liye ho sakta hai (desktop, mobile, library), Web API: Special type of API jo web ke through accessible hoti hai (browser, mobile apps).

Q4. Explain REST and SOAP Architecture. Mention shortcomings of SOAP.

Ans:- REST (Representational State Transfer): Lightweight architecture, HTTP methods (GET, POST, PUT, DELETE) use karta hai. JSON/XML format me data exchange hota hai, SOAP (Simple Object Access Protocol): Heavy protocol, XML format use karta hai, strict standards follow karta hai.
Shortcomings of SOAP: Complex implementation, Slower performance due to XML overhead, Less flexible compared to REST, Requires more bandwidth.

Q5. Differentiate between REST and SOAP.

Ans:- REST and SOAP are two different approaches for building web services, REST (Representational State Transfer) is a lightweight architecture that uses standard HTTP methods like GET, POST, PUT, and DELETE, It allows data to be exchanged in formats such as JSON or XML, making it faster, simpler, and more flexible, REST is widely used in modern applications because of its ease of use and efficiency, On the other hand, SOAP (Simple Object Access Protocol) is a heavier protocol that strictly relies on XML for data exchange and follows rigid standards, While SOAP provides strong security and reliability, it is more complex to implement and slower compared to REST, In short, REST is preferred for speed and simplicity, whereas SOAP is still used in enterprise systems where strict rules and high security are required.
